# Task 1 – Dataset Exploration & Cleaning

This notebook loads the original dataset, selects the required columns, performs basic cleaning, and saves a cleaned CSV file for later tasks.


```python
# 1. Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# For inline plots (if using Jupyter)
%matplotlib inline
```

```python
# 2. Load raw dataset
# Make sure this CSV file exists in the same folder as this notebook
csv_path = "SalesForCourse_quizz_table.csv"
df = pd.read_csv(csv_path)
print("Raw shape:", df.shape)
df.head()
```

```python
# 3. Select only the columns required for the model
cols_needed = ["Gender", "Age", "AnnualIncome", "SpendingScore"]

missing_cols = [c for c in cols_needed if c not in df.columns]
if missing_cols:
    print("WARNING: These expected columns are missing in the CSV:", missing_cols)

df_model = df[cols_needed].copy()
print("Filtered shape:", df_model.shape)
df_model.head()
```

```python
# 4. Check and handle missing values
print("Missing values per column:")
print(df_model.isnull().sum())

# Simple strategy: drop rows with any missing values in these columns
df_clean = df_model.dropna().copy()
print("Shape after dropping rows with missing values:", df_clean.shape)
```

```python
# 5. Descriptive statistics
df_clean.describe(include="all")
```

```python
# 6. Basic visualisations (optional but good for understanding)

# Age distribution
plt.figure()
df_clean["Age"].hist(bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")

# Annual income distribution
plt.figure()
df_clean["AnnualIncome"].hist(bins=20)
plt.title("Annual Income Distribution")
plt.xlabel("Annual Income")
plt.ylabel("Count")

# Spending score distribution
plt.figure()
df_clean["SpendingScore"].hist(bins=20)
plt.title("Spending Score Distribution")
plt.xlabel("Spending Score")
plt.ylabel("Count")

# Gender bar chart
plt.figure()
df_clean["Gender"].value_counts().plot(kind="bar")
plt.title("Gender Count")
plt.xlabel("Gender")
plt.ylabel("Count")
```

```python
# 7. Save cleaned dataset for later tasks
output_path = "cleaned_spending_data.csv"
df_clean.to_csv(output_path, index=False)
print("Saved cleaned data to:", output_path)
```
